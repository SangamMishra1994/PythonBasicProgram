### Explain the Pandas library and it's features?


#### What is Pandas?

Pandas is a Python library used for data manipulation and analysis, especially for structured data like tables (rows and columns).

In simple terms: Pandas helps you load, clean, transform, and analyze data easily.

#### Why Pandas is Needed
Working with raw data (CSV, Excel, JSON) using Python lists/dictionaries is difficult and inefficient.

Pandas provides:

- Table-like structure (like Excel)
- Fast operations (built on NumPy)
- Easy data cleaning and transformation



#### Core Data Structure
1. Series (1D)

```python
import pandas as pd

s = pd.Series([10, 20, 30])
print(s)
```

2. DataFrame - 2D array

```python
data = {
    "Name": ["A", "B"],
    "Marks": [10, 20]
}

output = pd.DataFrame(data)
print(output)
```

```python
data2 = {
    "marks": [75, 90],
    "Grade": ["c+", "A"]
}

print(pd.DataFrame(data2))
```
