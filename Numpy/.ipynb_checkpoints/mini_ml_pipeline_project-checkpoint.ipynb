```python
# Mini ML Pipeline Project
# Step - 1 Predict whether a student passes based on marks

import numpy as np

marks = np.array([35, 50, 65, 80, 90])
labels = np.array([0, 0, 1, 1, 1])  # 0=Fail, 1=Pass
```

```python
# Step 2- Normalize
mean = np.mean(marks)
std = np.std(marks)

marks_norm = (marks - mean) / std

print(f"Mean:- {mean} and marks norm = {marks_norm}")
```

```python
# Step - 3 Simple Model(Threshold)
predictions = np.where(marks_norm > 0, 1, 0)
```

```python
# Step 4 - Accuracy
accuracy = np.sum(predictions == labels) / len(labels)
print("Accuracy :- ", accuracy)
```

```python
# Step 5 - New Prediction
new_marks = 65
new_norm = (new_marks - mean) / std

result = 1 if new_norm > 0 else 0
print(new_norm)
print("Pass" if result else "Fail")
```
